```python
import pandas as pd
from numpy import nan
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Projetar-Solucoes/materiais-trilha/main/semana-04/solicitacoes_semana4.csv")
print(df.shape)
```

```python
df = df.drop_duplicates()                  # tira as linhas repetidas
display(df)
```

```python
df_dup = df.copy()
df_dup["categoria"] = df_dup["categoria"].str.strip().str.lower()
padrao = {
    "suporte ti": "Suporte TI",
    "suporte de ti": "Suporte TI",
    "acesso": "Acesso",
    "financeiro": "Financeiro",
    "infraestrutura": "Infraestrutura",
    "infra": "Infraestrutura",
    "dúvida": "Dúvida",
    "duvida": "Dúvida",
}

df_dup["categoria"] = df_dup["categoria"].map(padrao)
print(df_dup.shape)
```

```python
df_dup["status"] = df_dup["status"].str.capitalize() #quantidade de avaliações por categoria
df_dup['avaliacao'] = df_dup['avaliacao'].replace(7, nan)
media_avaliacao = df_dup.groupby("categoria")["avaliacao"].mean()
display(media_avaliacao)
display(df_dup['categoria'].value_counts())
```
